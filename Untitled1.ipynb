```python
import pandas as pd 
```

```python
June_lebaraToImbil = pd.read_csv ('/Users/ayobamisalami/Downloads/LEBARA-IMBIL CDR/CDR 01-30-June-2026.csv')
```

```python
July_lebaraToImbil = pd.read_csv ('/Users/ayobamisalami/Downloads/LEBARA-IMBIL CDR/CDR 01-31-July-2026.csv')
```

```python
August_lebaraToImbil = pd.read_csv('/Users/ayobamisalami/Downloads/LEBARA-IMBIL CDR/CDR 01-31-August-2026.csv')
```

```python
September_lebaraToImbil = pd.read_csv('/Users/ayobamisalami/Downloads/LEBARA-IMBIL CDR/CDR 01-28-September-2026.csv')
```

```python
lebaraToImbil = pd.concat(
    [
        June_lebaraToImbil,
        July_lebaraToImbil,
        August_lebaraToImbil,
        September_lebaraToImbil
    ],
    ignore_index=True
)
```

```python
lebaraToImbil.head(20)
```

```python
lebaraToImbil = lebaraToImbil.rename(columns={
    'Caller': 'A_Number',
    'Callee': 'B_Number'
})
```

```python
lebaraToImbil.head(20)
```

```python
distinct_A_Numbers = (
    lebaraToImbil[lebaraToImbil['Call duration'] != 0]
    [['A_Number']]
    .drop_duplicates()
    .reset_index(drop=True)
)
```
